# STCIS 33개 광역버스 노선 ID 조회

버스번호를 STCIS 버스노선 API로 조회하여 교통카드 합성데이터 API에 사용할 노선 ID를 가져온다.

- 경기도 시도코드: `41`
- 조회 대상: GTX-A 북측 33개 고유 버스번호
- STCIS 요청변수: `apikey`, `sdCd`, `routeNo`

```python
# 프로젝트 루트(UROP) 자동 탐색 — 절대경로 없이 어느 컴퓨터에서도 동작
import sys
from pathlib import Path
UROP_ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'config.py').is_file() and (p / 'code').is_dir())
for _d in [UROP_ROOT, *sorted(d for d in (UROP_ROOT / 'code').rglob('*') if d.is_dir() and d.name != '__pycache__')]:
    if str(_d) not in sys.path:
        sys.path.append(str(_d))

from pathlib import Path
from getpass import getpass
import json
import time
import requests
import pandas as pd

STCIS_URL = "https://stcis.go.kr/openapi/busroute.json"
STCIS_KEY = getpass("STCIS 인증키를 입력하세요: ").strip()
SD_CD = "41"

if not STCIS_KEY:
    raise ValueError("STCIS 인증키가 입력되지 않았습니다.")

print("STCIS 설정 완료")
```

### 셀 2. 32개 노선 CSV 불러오기\n

### ?? ? 2. ?? ?? CSV ??


```python
# 앞에서 추출한 33개 버스번호 불러오기
route_file = Path(UROP_ROOT / "data/raw/gtx_a/route_selection/gtx_a_north_33_unique_routes.csv")
routes = pd.read_csv(route_file, dtype=str, encoding="utf-8-sig")
routes["버스번호"] = routes["버스번호"].str.strip().str.lstrip("'")
routes = routes.drop_duplicates(subset=["버스번호"]).reset_index(drop=True)

print(f"조회 대상 버스번호: {len(routes)}개")
display(routes[["버스번호", "GTX_A_역", "대상지역"]])
```

### 셀 3. 노선번호로 STCIS 노선 ID 조회 함수 정의\n

### ?? ? 3. ????? STCIS ?? ?? ?? ?? ??


```python
def request_stcis_route(route_no):
    params = {
        "apikey": STCIS_KEY,
        "sdCd": SD_CD,
        "routeNo": route_no,
    }
    response = requests.get(STCIS_URL, params=params, timeout=30)
    response.raise_for_status()
    return response.json()

def walk_dicts(obj):
    if isinstance(obj, dict):
        yield obj
        for value in obj.values():
            yield from walk_dicts(value)
    elif isinstance(obj, list):
        for value in obj:
            yield from walk_dicts(value)

def find_route_records(payload):
    # 같은 버스번호가 여러 노선으로 등록될 수 있으므로
    # 노선ID·노선번호·기점·종점을 모두 후보별로 반환
    records = []
    for item in walk_dicts(payload):
        route_id = item.get("routeId", item.get("rteId", item.get("rte_id", item.get("routeID", ""))))
        route_no = item.get("routeNo", item.get("route_no", ""))
        start = item.get("stgSttnNma", item.get("stgSttnNm", item.get("startStation", "")))
        end = item.get("arrSttnNma", item.get("arrSttnNm", item.get("endStation", "")))
        if route_id not in (None, ""):
            record = {
                "STCIS_노선ID": str(route_id).strip(),
                "STCIS_노선번호": str(route_no).strip(),
                "기점": str(start).strip(),
                "종점": str(end).strip(),
            }
            if record not in records:
                records.append(record)
    return records

def mask_key(url):
    return url.split("apikey=")[0] + "apikey=***" if "apikey=" in url else url
```

### 셀 5. 전체 노선의 STCIS 후보 조회 및 저장\n

### ?? ? 5. ?? ?? ??? STCIS ?? ID ?? ? ??


```python
# 33개 버스번호 전체 조회
results = []
raw_responses = {}

for i, route_no in enumerate(routes["버스번호"], start=1):
    try:
        payload = request_stcis_route(route_no)
        raw_responses[route_no] = payload
        route_records = find_route_records(payload)
        if route_records:
            for record in route_records:
                results.append({
                    "버스번호": route_no,
                    "STCIS_노선ID": record["STCIS_노선ID"],
                    "STCIS_노선번호": record["STCIS_노선번호"],
                    "기점": record["기점"],
                    "종점": record["종점"],
                    "상태": "후보 확인",
                    "오류": "",
                })
            print(f"[{i}/{len(routes)}] {route_no} -> 후보 {len(route_records)}개")
        else:
            results.append({"버스번호": route_no, "STCIS_노선ID": "", "STCIS_노선번호": "", "기점": "", "종점": "", "상태": "노선ID 미확인", "오류": ""})
            print(f"[{i}/{len(routes)}] {route_no} -> 노선ID 미확인")
    except Exception as e:
        results.append({"버스번호": route_no, "STCIS_노선ID": "", "상태": "실패", "오류": repr(e)})
        print(f"[{i}/{len(routes)}] {route_no} -> 실패: {e}")
    time.sleep(0.2)

id_map = pd.DataFrame(results)
display(id_map)
```

### 셀 6. 후보 노선과 최종 노선 ID 비교\n

### ?? ? 6. ?? ?? ??? STCIS ?? ?? ??


```python
# 원본 CSV의 버스번호·기점·종점과 STCIS 후보 비교
candidate_file = Path(UROP_ROOT / "data/raw/gtx_a/route_selection/gtx_a_seoul_metropolitan_bus_candidates.csv")
expected = pd.read_csv(candidate_file, dtype=str, encoding="utf-8-sig")
expected = expected[expected["대상지역"].isin(["운정·파주", "일산서구", "덕양구"])].copy()
expected["버스번호"] = expected["버스번호"].str.strip().str.lstrip("'")

def norm_text(value):
    if pd.isna(value):
        return ""
    return "".join(str(value).split()).replace("'", "")

expected["비교_버스번호"] = expected["버스번호"].map(norm_text)
expected["비교_기점"] = expected["기점"].map(norm_text)
expected["비교_종점"] = expected["종점"].map(norm_text)
id_map["비교_버스번호"] = id_map["버스번호"].map(norm_text)
id_map["비교_기점"] = id_map["기점"].map(norm_text)
id_map["비교_종점"] = id_map["종점"].map(norm_text)

comparison = expected.merge(id_map, on="비교_버스번호", how="left", suffixes=("_원본", "_STCIS"))
comparison["비교결과"] = "후보 없음"
same_id = comparison["STCIS_노선ID"].notna() & comparison["STCIS_노선ID"].ne("")
same_start = comparison["비교_기점_원본"] == comparison["비교_기점_STCIS"]
same_endpoints = same_start & (comparison["비교_종점_원본"] == comparison["비교_종점_STCIS"])
# STCIS의 종점은 노선 등록 종점이고, 원본 CSV의 종점은 서울행 정류장일 수 있어
# 종점까지 같은 경우는 별도로 표시하되, 기점이 같은 후보도 확인할 수 있도록 분류
comparison.loc[same_id, "비교결과"] = "노선번호·기점 일치 후보"
comparison.loc[same_id & same_endpoints, "비교결과"] = "노선번호·기종점 일치"

show_cols = ["GTX_A_역", "대상지역", "버스번호", "기점", "종점", "STCIS_노선ID", "STCIS_노선번호", "기점_STCIS", "종점_STCIS", "비교결과"]
print("비교 대상 원본 역-노선 조합:", len(expected))
print("노선번호·기종점 일치 조합:", (comparison["비교결과"] == "노선번호·기종점 일치").sum())
print("노선번호·기점 일치 후보 조합:", (comparison["비교결과"] == "노선번호·기점 일치 후보").sum())
print("후보 없음 조합:", (comparison["비교결과"] == "후보 없음").sum())
# 실제로 확인할 후보만 표시: 버스번호와 기점이 일치하는 행
matched_candidates = comparison[same_id & same_start].copy()
display(matched_candidates[[c for c in show_cols if c in matched_candidates.columns]])
```
